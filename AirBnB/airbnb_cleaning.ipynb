```python
import pandas as pd
```

```python
data = pd.read_csv("Airbnb_Open_Data.csv")
```

```python
data.info()
```

# Checklist For Data Cleaning

1. Deleting Redunant Columns
2. Renaming Columns
3. Dropping Duplicates
4. Removing NaN Values
5. Cleaning Individual Columns
6. Checking for Extra Transforms

# 1: Dropping Redundant Columns

```python
data.columns
```

```python
columns_to_keep = ['NAME', 'host id', 'host_identity_verified', 'host name',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy', 'room type',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'number of reviews', 'last review']
columns_to_drop = ['reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license', 'id']
```

```python
len(columns_to_keep)
```

```python
len(columns_to_drop)
```

```python
data.drop(columns = columns_to_drop, inplace = True)
```

# 2: Renaming the Columns

```python
data.columns
```

```python
data.rename(columns= {"NAME": "Name"}, inplace = True)
```

```python
new_column_names = []
for i in data.columns:
    new_column_names.append(i.upper())
```

```python
data.columns = new_column_names
```

```python
data
```

# 3: Dropping Duplicates

```python
data.drop_duplicates(inplace= True)
```

```python
data.duplicated().sum()
```

# 4: Removing NaN Values

```python
data.isna().sum()
```

```python
data.drop(columns=["LAST REVIEW"], inplace=True)
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

# 5th Method, Cleaning Individual Columns

```python
data["HOST_IDENTITY_VERIFIED"] = data["HOST_IDENTITY_VERIFIED"].str.upper()
```

```python
data["INSTANT_BOOKABLE"] = data["INSTANT_BOOKABLE"].apply(lambda x:1 if x == True else 0)
```

```python
data.reset_index(inplace=True)
```

```python
data
```

# 6th Method, Check for Transformations

```python
type(data["PRICE"][15])
```

```python
data["PRICE"].head(3)
```

```python
data["PRICE"] = data["PRICE"].str.replace("$", "")
```

```python
data["PRICE"] = data["PRICE"].str.replace(",", "")
```

```python
data["PRICE"] = data["PRICE"].str.replace(" ", "")
```

```python
data["PRICE"] = data["PRICE"].astype(int)
```

```python
data["PRICE"][1]
```

```python
data.to_csv("Cleaned_Data.csv", index=False)
```
