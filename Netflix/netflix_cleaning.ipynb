# Getting Rid of Nulls

```python
data.isnull().sum()
```

```python
data['director'] = data['director'].fillna('N/A')
data['cast'] = data['cast'].fillna('N/A')
data['country'] = data['country'].fillna('N/A')
```

```python
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

# Treat Any Duplicates

```python
data.duplicated().sum()
```

# Cleaning Columns

```python
data.info()
```

```python
data['show_id'].head(3)
```

```python
data['show_id'] = data['show_id'].str.replace('s', '', regex=True)
data['show_id'] = data['show_id'].astype(int)
```

```python
data['show_id'].head(3)
```

```python
data['date_added'].head(5)
```

```python
data['date_added'] = data['date_added'].str.strip()
```

```python
data['date_added'] = data['date_added'].astype('datetime64[ns]')
```

```python
data['date_added'].head(5)
```

# Export Cleaned Table

```python
data.reset_index(drop=True)
```

```python
data.to_csv("Cleaned_Netflix_Data.csv", index=False)
```
